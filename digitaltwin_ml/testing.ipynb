```python
import pandas as pd
import joblib
```

```python
residual_model = joblib.load("dt_residual_model.pkl")
```

```python
def digital_twin_baseline(input_data):
    return (
        700
        + 200 * input_data["seasonal_stress"]
        + 150 * input_data["harvest_index"]
    )
```

```python
input_data = {
    "seasonal_stress": 0.85,
    "LAI_max": 5.4,
    "leaf_growth_rate": 0.09,
    "RUE": 1.50,
    "harvest_index": 0.38
}
```

```python
X_input = pd.DataFrame([input_data])
```

```python
dt_pred = digital_twin_baseline(input_data)
```

```python
residual_correction = residual_model.predict(X_input)[0]
```

```python
final_yield = dt_pred + residual_correction
final_yield
```
