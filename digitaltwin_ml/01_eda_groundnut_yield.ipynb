```python
X = df[['area_ha', 'temp_c', 'humidity_pct', 'rainfall_mm', 'solar_wm2']]
y = df['yield_kg_per_ha']
```

```python
X_train = X[df['Year'] <= 2018]
X_test  = X[df['Year'] > 2018]

y_train = y[df['Year'] <= 2018]
y_test  = y[df['Year'] > 2018]
```

```python
X_train.shape, X_test.shape
```

```python
%pip install scikit-learn

```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
y_pred_lr = lr.predict(X_test)
```

```python
%pip install -U scikit-learn
```

```python
import numpy as np

mae = mean_absolute_error(y_test, y_pred_lr)
mse = mean_squared_error(y_test, y_pred_lr)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred_lr)

mae, rmse, r2

```

```python
pd.Series(lr.coef_, index=X.columns)
```

```python
lr.intercept_
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf = RandomForestRegressor(
    n_estimators=200,
    random_state=42
)
```

```python
rf.fit(X_train, y_train)
```

```python
y_pred_rf = rf.predict(X_test)
```

```python
mae_rf = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = mean_squared_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)

mae_rf, rmse_rf, r2_rf
```

```python
rf_ctrl = RandomForestRegressor(
    n_estimators=300,
    max_depth=3,           # KEY FIX
    min_samples_leaf=2,    # KEY FIX
    random_state=42
)
```

```python
rf_ctrl.fit(X_train, y_train)
```

```python
y_pred_rf_ctrl = rf_ctrl.predict(X_test)
```

```python
mae_rf_c = mean_absolute_error(y_test, y_pred_rf_ctrl)
rmse_rf_c = mean_squared_error(y_test, y_pred_rf_ctrl)
r2_rf_c = r2_score(y_test, y_pred_rf_ctrl)

mae_rf_c, rmse_rf_c, r2_rf_c
```

```python
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
svr_model = Pipeline([
    ("scaler", StandardScaler()),
    ("svr", SVR(kernel="rbf", C=10, epsilon=0.1, gamma="scale"))
])
```

```python
svr_model.fit(X_train, y_train)
```

```python
y_pred_svr = svr_model.predict(X_test)
```

```python
mae_svr = mean_absolute_error(y_test, y_pred_svr)
rmse_svr = mean_squared_error(y_test, y_pred_svr)
r2_svr = r2_score(y_test, y_pred_svr)

mae_svr, rmse_svr, r2_svr
```
